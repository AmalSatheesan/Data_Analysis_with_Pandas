Read the `imdb_top_1000.csv` file with the following columns: 

`['Series_Title' ,'IMDB_Rating', 'No_of_Votes', 'Gross', 'Released_Year', 'Genre']`



```python
cols = ['Series_Title' ,'IMDB_Rating', 'No_of_Votes', 'Gross', 'Released_Year', 'Genre']
movies= pd.read_csv('imdb_top_1000.csv' , usecols = cols)
movies 
```

```python
movies.info()
```

```python
movies.describe()
```

```python
movies['Gross'] = movies['Gross'].str.replace(',', '')
```

```python
movies['Gross'].fillna(0,inplace=True)
```

```python
movies['Gross'] = movies['Gross'].astype('int64')
```

```python
movies.describe(percentiles=[.20,.40,.60,.80]).round()
```

```python
movies.count()
```

```python
movies.mean(numeric_only=True).round()
```

```python
movies.min(numeric_only=True).round()
```

```python
movies.max(numeric_only=True).round()
```

```python
movies.quantile(q=.5,numeric_only=True).round(2)
```

```python
movies.describe(include='O')
```

```python
movies.nlargest(n=10,columns=['IMDB_Rating','No_of_Votes'])
```

```python
movies.nsmallest(n=10,columns=['IMDB_Rating','No_of_Votes']) #keep is used to keep duplicate or not.
```

```python
movies.set_index('Series_Title', drop=False, inplace =False)
```

```python
movies.reset_index( drop=True, inplace =False)
```

```python
movies
```

Find the lowest scored movie with the lowest amount of votes

```python
movies.nsmallest(n=1 , columns= ['IMDB_Rating','No_of_Votes'])
```

Find the highest scored movie with the highest amount of votes

```python
movies.nlargest(n=1 , columns= ['IMDB_Rating','No_of_Votes'])
```

Find the top 5 genres

```python
movies.value_counts('Genre').head(5)
```
